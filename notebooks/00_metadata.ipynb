# Metadata

Summary counts used in the Methods section of the paper. Cells are grouped by topic; all outputs are printed only (no files saved).

## Setup

```python
import pandas as pd
```

```python
df_filtered        = pd.read_csv('../data/filtered_observations.csv')
df_predator_speeds = pd.read_csv('../data/predator_attack_speeds.csv')
df_all             = pd.read_csv('../data/all_observations.csv')
```

## Study videos

Total number of recorded trials, split by whether a predator attack response was observed.

```python
count_videos              = df_filtered['Video_ID'].nunique()
count_response_videos     = df_filtered.loc[df_filtered['Response_Video'] == 'Yes', 'Video_ID'].nunique()
count_nonresponse_videos  = df_filtered.loc[df_filtered['Response_Video'] == 'No',  'Video_ID'].nunique()

print('Number of videos:              ', count_videos)
print('Number of response videos:     ', count_response_videos)
print('Number of non-response videos: ', count_nonresponse_videos)
```

## Analysis dataset

Counts from `filtered_observations.csv`: Brown Chromis and Bicolored Damselfish only, FR and NRL_nr response categories, viewing angle < 55°, body size < 500 mm.

```python
count_FR     = (df_filtered['Response_Category'] == 'FR').sum()
count_NRL_nr = (df_filtered['Response_Category'] == 'NRL_nr').sum()
count_total  = count_FR + count_NRL_nr

print('Number of first responders (FR):                ', count_FR)
print('Number of non-responding fish with LoS (NRL-NR):', count_NRL_nr)
print('Total fish in analysis dataset:                 ', count_total)
```

## Predator attack speeds

Number of Bar Jack (*Caranx ruber*) attacks recorded in the same reef habitat, used to characterise natural predator attack speeds.

```python
count_jack_speeds = df_predator_speeds['TrackID'].nunique()
print('Number of predator attack speeds recorded:', count_jack_speeds)
```

## Methods justifications

### Sensory-motor delay

The minimum time after the first responder (FR) for fish categorised as SRNL (secondary responders with no line of sight). Used to justify the 5-frame sensory-motor delay threshold.

```python
SRNL = df_all.loc[df_all['Response_Category'] == 'SRNL']
print('Minimum time after FR for SRNL fish (frames):', int(min(SRNL['After_FR'])))
```

### Viewing angle cutoff (55°)

Fish with a viewing angle > 55° cannot reliably see the stimulus (reflective screen). This checks how many such fish existed and whether any were classified as FR — justifying their exclusion from the analysis.

```python
outside_angle    = df_all.loc[df_all['Viewing_Angle'] > 55]
fr_outside_angle = outside_angle.loc[outside_angle['Response_Category'] == 'FR']

print('Fish with viewing angle > 55°:', len(outside_angle))
print('Of which classified as FR:    ', len(fr_outside_angle))
print()
print('Viewing angles of FR fish outside cutoff (degrees):')
print(fr_outside_angle['Viewing_Angle'].round(1).values)
print()
print('Distance to stimulus of FR fish outside cutoff (m):')
print((fr_outside_angle['Distance_to_Stimulus'] / 1000).round(2).values)
```
